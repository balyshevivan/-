```python
# Импорт библиотек
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Настройка отображения графиков
%matplotlib inline

# Загрузка датасета
df = pd.read_csv('train.csv')

# Первый взгляд на данные
df.head()
```

```python
# Количество строк и столбцов
print("Размер датасета (строки, столбцы):", df.shape)

# Названия всех столбцов
print("\nСтолбцы:")
print(df.columns.tolist())

# Типы данных столбцов
print("\nТипы данных:")
print(df.dtypes)

# Дополнительно: информация о пропусках
print("\nПропущенные значения:")
print(df.isnull().sum())
```

```python
# Количество пассажиров каждого класса
print("Пассажиры по классам:")
print(df['Pclass'].value_counts().sort_index())

# Количество мужчин и женщин
print("\nПассажиры по полу:")
print(df['Sex'].value_counts())

# Разделение на возрастные группы
bins = [0, 12, 18, 30, 50, 80]
labels = ['Дети (0-12)', 'Подростки (13-18)', 'Молодые (19-30)',
          'Взрослые (31-50)', 'Пожилые (51+)']
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)

print("\nВозрастные группы:")
print(df['AgeGroup'].value_counts().sort_index())

# Средний возраст
print("\nСредний возраст:", round(df['Age'].mean(), 2))

# Медианный возраст
print("Медианный возраст:", df['Age'].median())

# Средняя стоимость билета
print("Средняя стоимость билета:", round(df['Fare'].mean(), 2))
```

```python
# Средняя стоимость билета по классам
print("Средняя стоимость билета по классам:")
print(df.groupby('Pclass')['Fare'].mean().round(2))

# Средний возраст по классам
print("\nСредний возраст по классам:")
print(df.groupby('Pclass')['Age'].mean().round(2))

# Количество мужчин и женщин в каждом классе
print("\nПол по классам:")
print(df.groupby(['Pclass', 'Sex']).size().unstack())

# В каком классе больше всего пассажиров
print("\nСамый многочисленный класс:",
      df['Pclass'].value_counts().idxmax())

# Самый дорогой билет
max_fare_row = df.loc[df['Fare'].idxmax()]
print("\nСамый дорогой билет:")
print(max_fare_row[['Name', 'Pclass', 'Age', 'Sex', 'Survived', 'Fare']])
```

```python
# Общее количество выживших и погибших
print("Выжившие/погибшие:")
print(df['Survived'].value_counts().rename({0: 'Погиб', 1: 'Выжил'}))

# Выживаемость по полу
print("\nВыживаемость по полу (доля):")
print(df.groupby('Sex')['Survived'].mean().round(3))

# Выживаемость по полу и классу
print("\nВыживаемость по классу и полу:")
print(df.groupby(['Pclass', 'Sex'])['Survived'].mean().round(3).unstack())

# Группа с наибольшей долей выживших
best = df.groupby(['Pclass', 'Sex'])['Survived'].mean().idxmax()
best_val = df.groupby(['Pclass', 'Sex'])['Survived'].mean().max()
print(f"\nНаибольшая доля выживших: класс {best[0]}, пол {best[1]} — {best_val:.1%}")
```

```python
# График количества пассажиров по классам
plt.figure(figsize=(7, 4))
df['Pclass'].value_counts().sort_index().plot(
    kind='bar', color=['gold', 'silver', '#cd7f32'], edgecolor='black')
plt.title('Количество пассажиров по классам')
plt.xlabel('Класс')
plt.ylabel('Количество пассажиров')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

Вопрос: Влияло ли наличие родственников (SibSp + Parch) на выживаемость пассажиров?

```python
# Создаём признак "Размер семьи"
df['FamilySize'] = df['SibSp'] + df['Parch']
df['FamilyCat'] = df['FamilySize'].apply(
    lambda x: 'Один' if x == 0 else ('Малая семья (1-3)' if x <= 3 else 'Большая семья (4+)'))

print("Выживаемость в зависимости от размера семьи:")
print(df.groupby('FamilyCat')['Survived'].agg(['mean', 'count']).round(3))
```

пассажиры, путешествовавшие в одиночку, выживали в 30% случаев; пассажиры с небольшой семьёй (1–3 родственника) — в 58%; а большие семьи (4+) — лишь в 16%. То есть наличие небольшой семьи повышало шансы на спасение (вероятно, помогали друг другу), а большие семьи, наоборот, снижали шансы — вероятно, из-за сложности эвакуации и принадлежности к 3-му классу.
